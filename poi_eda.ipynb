```python
fig, ax = plt.subplots(figsize=(10,10))
gdf.plot(
    ax=ax,
    column="station_id",
    categorical=True,
    legend=True,
    markersize=10,
    alpha=0.7
)
plt.title("POI Locations by Station")
plt.show()
```
